```python
import random
import numpy as np
import matplotlib.pyplot as plt

num_rolls = int(input("Enter the number of dice rolls: "))

fair_rolls = [random.randint(1, 6) for _ in range(num_rolls)]

biased_probs = [0.1, 0.1, 0.05, 0.1, 0.5, 0.15]
biased_rolls = np.random.choice([1, 2, 3, 4, 5, 6],
                                 size=num_rolls,
                                 p=biased_probs)

fair_counts = np.bincount(fair_rolls, minlength=7)[1:]
biased_counts = np.bincount(biased_rolls, minlength=7)[1:]

fair_observed_probs = fair_counts / num_rolls
biased_observed_probs = biased_counts / num_rolls

fair_expected_prob = [1/6] * 6
biased_expected_prob = biased_probs

print("\nFAIR DICE:")
print(f"{'Face':<5} {'Expected Prob':<15} {'Observed Prob'}")
for i in range(6):
    print(f"{i+1:<5} {fair_expected_prob[i]:<15.4f} {fair_observed_probs[i]:.4f}")

print("\nBIASED DICE:")
print(f"{'Face':<5} {'Expected Prob':<15} {'Observed Prob'}")
for i in range(6):
    print(f"{i+1:<5} {biased_expected_prob[i]:<15.4f} {biased_observed_probs[i]:.4f}")

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.bar(range(1, 7), fair_observed_probs, color='darkblue', alpha=0.8, label='Observed')
plt.bar(range(1, 7), fair_expected_prob, color='lightblue', alpha=0.5, label='Expected')
plt.xticks(range(1, 7))
plt.title("Fair Dice Rolls_AMIR BHATTARAI")
plt.xlabel("Dice Face")
plt.ylabel("Probability")
plt.legend()

plt.subplot(1, 2, 2)
plt.bar(range(1, 7), biased_observed_probs, color='purple', alpha=0.7, label='Observed')
plt.bar(range(1, 7), biased_expected_prob, color='orange', alpha=0.3, label='Expected')
plt.xticks(range(1, 7))
plt.title("Biased Dice Rolls_AMIR BHATTARAI")
plt.xlabel("Dice Face")
plt.ylabel("Probability")
plt.legend()

plt.tight_layout()
plt.show()
```
