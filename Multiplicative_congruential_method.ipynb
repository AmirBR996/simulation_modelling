```python
# Function to generate random numbers using Multiplicative Congruential Method
def multiplicative_congruential_method(X0, a, m, n):
    Xn = X0  # Initialize seed value
    print("\nGenerated Random Numbers [AMIR_BHATTARAI ]:")
    for i in range(n):
        Xn = (a * Xn) % m  # MCM formula (no increment term)
        print(f"X{i} = {Xn}")  # Print the generated number

# Taking user input
X0 = int(input("Enter seed value (X0): "))
a = int(input("Enter multiplier (a): "))
m = int(input("Enter modulus (m): "))
n = int(input("Enter number of random numbers to generate: "))

# Generate and display random numbers
multiplicative_congruential_method(X0, a, m, n)1
```
